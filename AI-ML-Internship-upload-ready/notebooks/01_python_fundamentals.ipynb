## 1. Variables and data types

```python
name = "Kashif"
age = 21
gpa = 8.4
is_intern = True
for v in (name, age, gpa, is_intern):
    print(f"{v!r:>10}  ->  {type(v).__name__}")
```

## 2. Operators

```python
a, b = 17, 5
print("a + b =", a + b, "| a - b =", a - b, "| a * b =", a * b)
print("a / b =", a / b, "| a // b =", a // b, "| a % b =", a % b, "| a ** b =", a ** b)
print("a > b and b > 0 ->", a > b and b > 0)
print("a == 17 or b == 0 ->", a == 17 or b == 0)
```

## 3. Conditional statements

```python
def grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    elif score >= 60:
        return "C"
    return "F"

for s in (95, 80, 62, 40):
    print(s, "->", grade(s))
```

## 4. Loops

```python
# for loop with range
print([i ** 2 for i in range(1, 8)])

# while loop
n, steps = 27, 0
while n != 1:                      # Collatz sequence
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print("Collatz steps for 27:", steps)

# loop control
for i in range(10):
    if i == 3:
        continue
    if i == 6:
        break
    print(i, end=" ")
```

## 5. Functions

```python
def stats(numbers):
    """Return mean, median and mode of a list of numbers."""
    s = sorted(numbers)
    mean = sum(s) / len(s)
    mid = len(s) // 2
    median = s[mid] if len(s) % 2 else (s[mid - 1] + s[mid]) / 2
    mode = max(set(s), key=s.count)
    return mean, median, mode

print(stats([72, 85, 85, 90, 64, 78]))

def is_prime(n):
    return n > 1 and all(n % d for d in range(2, int(n ** 0.5) + 1))

print([n for n in range(2, 40) if is_prime(n)])
```

## 6. Lists, tuples, dictionaries and sets

```python
# List - ordered, mutable
marks = [72, 85, 90, 64, 78]
marks.append(88); marks.sort(reverse=True)
print("List  :", marks, "| top 3:", marks[:3])

# Tuple - ordered, immutable
point = (12.5, 7.25)
x, y = point
print("Tuple :", point, "-> x =", x, ", y =", y)

# Dictionary - key/value
student = {"name": "Asha", "branch": "CSE", "marks": {"math": 91, "ml": 88}}
student["marks"]["dsa"] = 95
print("Dict  :", student)
print("Average:", sum(student["marks"].values()) / len(student["marks"]))

# Set - unique, unordered
a, b = {"python", "sql", "git"}, {"git", "docker", "python"}
print("Set   : union", a | b, "| intersection", a & b, "| difference", a - b)
```

## 7. Comprehensions

```python
words = ["data", "model", "python", "git", "notebook"]
print({w: len(w) for w in words})
print([w.upper() for w in words if len(w) > 4])
```

## 8. File handling

```python
import csv, os

rows = [("name", "score"), ("Asha", 91), ("Ravi", 78), ("Meera", 85)]
with open("sample_scores.csv", "w", newline="") as f:
    csv.writer(f).writerows(rows)

with open("sample_scores.csv") as f:
    for row in csv.DictReader(f):
        print(row)

with open("notes.txt", "w") as f:
    f.write("Line one\nLine two\n")
with open("notes.txt") as f:
    print(f.read())

os.remove("sample_scores.csv"); os.remove("notes.txt")   # tidy up
```

## 9. Error handling

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        return float("nan")

print(safe_divide(10, 2), safe_divide(1, 0))
```
